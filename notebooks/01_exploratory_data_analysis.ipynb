Notebook n?y th?c hi?n c?c b??c ph?n t?ch s? b? b? d? li?u **ViWikiFC**:
- T?i v? ki?m tra c?c t?p d? li?u: `train.csv`, `dev.csv`, `test.csv`
- Ph?n ph?i nh?n ki?m ch?ng (`gold_label`)
- ?? d?i c?a ph?t bi?u (`claim`), b?ng ch?ng (`evidence`) v? ng? c?nh (`context`)
- Kh?o s?t c?c v? d? ?i?n h?nh cho t?ng nh?n: `Supports`, `Refutes`, `Not_Enough_Information`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Thi?t l?p hi?n th?
pd.set_option('display.max_colwidth', None)
plt.style.use('ggplot')
```

## 1. T?i c?c t?p d? li?u ViWikiFC

```python
train_df = pd.read_csv('../data/raw/train.csv')
dev_df = pd.read_csv('../data/raw/dev.csv')
test_df = pd.read_csv('../data/raw/test.csv')

print(f"K?ch th??c t?p Train: {train_df.shape}")
print(f"K?ch th??c t?p Dev:   {dev_df.shape}")
print(f"K?ch th??c t?p Test:  {test_df.shape}")
train_df.head(2)
```

## 2. Ph?n ph?i nh?n (Label Distribution)

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, df) in zip(axes, [('Train', train_df), ('Dev', dev_df), ('Test', test_df)]):
    df['gold_label'].value_counts().plot(kind='bar', ax=ax, color=['#2ca02c', '#d62728', '#1f77b4'])
    ax.set_title(f'Ph?n ph?i nh?n - {name}')
    ax.set_ylabel('S? l??ng m?u')
    ax.set_xticklabels(ax.get_xticklabels(), rotation=30)

plt.tight_layout()
plt.show()
```

## 3. Th?ng k? ?? d?i t? (Word Length Statistics)

```python
train_df['claim_words'] = train_df['claim'].astype(str).apply(lambda x: len(x.split()))
train_df['evidence_words'] = train_df['evidence'].astype(str).apply(lambda x: len(x.split()))

print("--- Th?ng k? ?? d?i t? c?a Claim ---")
print(train_df['claim_words'].describe())

print("\n--- Th?ng k? ?? d?i t? c?a Evidence ---")
print(train_df['evidence_words'].describe())
```

## 4. Xem m?u c? th? cho t?ng nh?n ki?m ch?ng

```python
for label in ['Supports', 'Refutes', 'Not_Enough_Information']:
    sample = train_df[train_df['gold_label'] == label].iloc[0]
    print(f"==================== [{label.upper()}] ====================")
    print(f"Claim:    {sample['claim']}")
    print(f"Evidence: {sample['evidence']}")
    print(f"Context:  {sample['context'][:200]}...")
    print()
```
