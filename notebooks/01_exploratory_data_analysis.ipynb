# 01. Khám Phá Dữ Liệu ViWikiFC (Exploratory Data Analysis)

Notebook này thực hiện các bước phân tích khám phá dữ liệu **ViWikiFC**:
- Kiểm tra kích thước các tập dữ liệu: `train.csv`, `dev.csv`, `test.csv`
- Phân tích phân phối nhãn kiểm chứng (`gold_label`)
- Phân tích độ dài phát biểu (`claim`) và bằng chứng (`evidence`)
- Khảo sát các mẫu ví dụ thực tế cho 3 nhãn: `Supports`, `Refutes`, `Not_Enough_Information`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập hiển thị bảng và đồ thị
pd.set_option('display.max_colwidth', None)
plt.style.use('ggplot')
```

## 1. Tải Các Tập Dữ Liệu ViWikiFC

```python
train_df = pd.read_csv('../data/raw/train.csv')
dev_df = pd.read_csv('../data/raw/dev.csv')
test_df = pd.read_csv('../data/raw/test.csv')

print(f"Tập Train: {train_df.shape}")
print(f"Tập Dev:   {dev_df.shape}")
print(f"Tập Test:  {test_df.shape}")
train_df.head(2)
```

## 2. Phân Phối Nhãn Kiểm Chứng (Label Distribution)

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, df) in zip(axes, [('Train', train_df), ('Dev', dev_df), ('Test', test_df)]):
    counts = df['gold_label'].value_counts()
    counts.plot(kind='bar', ax=ax, color=['#2ca02c', '#d62728', '#1f77b4'])
    ax.set_title(f'Phân phối nhãn - {name}')
    ax.set_ylabel('Số lượng mẫu')
    ax.set_xticklabels(ax.get_xticklabels(), rotation=30)

plt.tight_layout()
plt.show()
```

## 3. Thống Kê Độ Dài Từ (Word Length Distribution)

```python
train_df['claim_words'] = train_df['claim'].astype(str).apply(lambda x: len(x.split()))
train_df['evidence_words'] = train_df['evidence'].astype(str).apply(lambda x: len(x.split()))

print("--- Thống kê số lượng từ của Claim ---")
print(train_df['claim_words'].describe())

print("\n--- Thống kê số lượng từ của Evidence ---")
print(train_df['evidence_words'].describe())
```

## 4. Xem Các Mẫu Ví Dụ Tiêu Biểu Theo Từng Nhãn

```python
for label in ['Supports', 'Refutes', 'Not_Enough_Information']:
    sample = train_df[train_df['gold_label'] == label].iloc[0]
    print(f"==================== [{label.upper()}] ====================")
    print(f"Phát biểu (Claim):  {sample['claim']}")
    print(f"Bằng chứng (Evid): {sample['evidence']}")
    print(f"Ngữ cảnh (Context): {sample['context'][:180]}...")
    print()
```
